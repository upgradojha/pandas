```python

import pandas as pd

# create students dataframe
students = pd.DataFrame({
    "student_id": [101, 102, 103, 104, 105, 106],
    "name": ["aman", "priya", "rahul", "neha", "arjun", "sneha"],
    "age": [20, 21, 19, 20, 22, 21],
    "city": ["delhi", "ghaziabad", "noida", "delhi", "meerut", "noida"]
})

# create marks dataframe
marks = pd.DataFrame({
    "student_id": [101, 102, 103, 104, 105, 107],
    "subject": ["python"] * 6,
    "marks": [85, 92, 67, 78, 88, 72]
})

# create attendance dataframe
attendance = pd.DataFrame({
    "student_id": [101, 102, 103, 104, 106, 108],
    "attendance": [90, 95, 75, 82, 98, 60]
})

# create june students dataframe
students_june = pd.DataFrame({
    "student_id": [107, 108, 109, 110],
    "name": ["rohit", "anjali", "karan", "riya"],
    "age": [20, 19, 22, 21],
    "city": ["ghaziabad", "delhi", "meerut", "noida"]
})

# export all dataframes as csv files
students.to_csv("students.csv", index=False)
marks.to_csv("marks.csv", index=False)
attendance.to_csv("attendance.csv", index=False)
students_june.to_csv("students_june.csv", index=False)

print("all csv files created successfully")

# verify the exported files
print("\nfiles created:")
print("./data_stu/students.csv")
print("./data_stu/marks.csv")
print("./data_stu/attendance.csv")
print("./data_stu/students_june.csv")
```
