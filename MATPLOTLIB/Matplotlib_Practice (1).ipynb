```python
import matplotlib.pyplot as plt
import numpy as np
import matplotlib.pyplot as plt
from cycler import cycler


# Sophisticated dark-mode palette
nord_colors = ['#88c0d0', '#81a1c1', '#b48ead', '#a3be8c', '#ebcb8b']

plt.rcParams['axes.prop_cycle'] = cycler(color=nord_colors)



# Canteen sales (cups per month)
months = ["Jan", "Feb", "Mar", "Apr", "May", "Jun"]
chai   = [120, 135, 160, 150, 190, 210]
coffee = [80, 85, 90, 110, 95, 100]

# Daily phone screen time (hours)
apps  = ["Social", "Videos", "Games", "Music", "Study"]
hours = [2.5, 1.8, 0.9, 0.6, 1.2]

# A T20 match: runs scored in each of the 20 overs
team_a = [6, 8, 4, 12, 9, 7, 5, 6, 10, 8, 4, 7, 9, 14, 6, 11, 8, 13, 16, 18]
team_b = [10, 12, 9, 11, 8, 10, 6, 5, 7, 4, 6, 5, 8, 6, 7, 9, 10, 8, 11, 12]
```

```python
# Worm chart
overs = range(1, 21)
plt.plot(overs, np.cumsum(team_a), label = 'Team A', marker = '.', color = 'red', linestyle = '--')
plt.plot(overs, np.cumsum(team_b), label = 'Team b',marker = 'o', color = 'orange')
plt.xlabel('Overs on x-axis')
plt.ylabel('Cummulative sums of the runs of team')
plt.legend()
plt.grid(alpha = .2)
plt.show()
```

**Q6. Sorted screen time** - Draw a horizontal bar chart of screen time, sorted so the biggest bar is at the top.

*Hint:* sorted(zip(hours, apps)) sorts by hours

```python
vibrant_colors = ["#FFFB00", '#FF00FF', '#39FF14', "#FF5F1F", "#0044FFFF"]
sorted_data = sorted(zip(hours, apps))
sorted_hours, sorted_apps = zip(*sorted_data)
plt.grid(alpha = .2)
plt.barh(sorted_apps, sorted_hours, color =vibrant_colors)
plt.title("Daily Screen Time")
plt.ylabel('Hours')
plt.xlabel('Apps used')
plt.show()
```

**Q14. Label the hottest month** - Plot Delhi temperatures with dots and point an arrow at the hottest month, saying "Hottest: 33°C".

*Hint:* max() finds the top value, .index() its position

```python
plt.plot(month_names, delhi, marker='.', ls= '--', label = 'temp of delhi each month')

max_temp = max(delhi)
max_idx = delhi.index(max_temp)

plt.annotate('Best Month', xy=(max_idx, max_temp),
    xytext=(max_idx, max_temp-7),
             arrowprops={'arrowstyle': '->'})
# print(delhi)
plt.xlabel('Months name')
plt.ylabel('Temperature')
plt.legend()
plt.title("Hottest weather of delhi")
plt.savefig('./Hottest weather ', dpi = 1200, bbox_inches = "tight")
plt.show()
```
